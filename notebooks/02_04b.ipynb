## Working with Missing Data in Pandas

```python
import numpy as np
import pandas as pd

from pandas import DataFrame
```

```python
data = {
  'names' : ["Adam","Bob","Charles","Daisy","Edward","Fiona","Graham"],
  'age' : [20,21,23,19,29,27,26],
  'gender' : ['M','M','M','F','M','F','M'],
  'rank' : [4,2,6,1,3,5,7]
}
ranking_df = DataFrame(data)
ranking_df.iloc[3,:] = np.nan
ranking_df.iloc[2:5,1] = np.nan 
ranking_df.iloc[3:6,3] = np.nan
```

```python
ranking_df
```

```python
ranking_df.isna()
ranking_df.notna()
ranking_df.isnull()
ranking_df.notnull()
```

```python
ranking_df[pd.isnull(ranking_df['age'])]
```

```python
type(pd.isnull(ranking_df['age']))
```

```python
ranking_df.fillna('Not Missing')
```

```python
#The method argument is demised now, use ffill or bfill instead
#ranking_df.fillna(method='pad')
ranking_df.ffill()
```

```python
ranking_df.bfill()
```

```python
ranking_df.iloc[:,[1,3]].interpolate(method='linear')
```

```python
ranking_df.dropna()
```

```python
ranking_df.dropna(how='all')
```

```python
ranking_df.dropna(axis=1,how='all')
```
